```python
import os
import cv2

IMG_SIZE = 224

def preprocess_folder(folder_path):
    for class_name in os.listdir(folder_path):
        class_path = os.path.join(folder_path, class_name)

        if not os.path.isdir(class_path):
            continue

        for img_name in os.listdir(class_path):
            img_path = os.path.join(class_path, img_name)

            img = cv2.imread(img_path)
            if img is None:
                os.remove(img_path)
                continue

            img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
            cv2.imwrite(img_path, img)

# Run preprocessing
preprocess_folder(os.path.join(DATA_PATH, "train"))
preprocess_folder(os.path.join(DATA_PATH, "test"))

print("Preprocessing Completed ✅")
```

```python
import shutil
import random

train_path = os.path.join(DATA_PATH, "train")
val_path = os.path.join(DATA_PATH, "val")

os.makedirs(val_path, exist_ok=True)

for class_name in os.listdir(train_path):
    class_dir = os.path.join(train_path, class_name)

    if not os.path.isdir(class_dir):
        continue

    images = os.listdir(class_dir)
    random.shuffle(images)

    split_size = int(0.15 * len(images))

    val_class_dir = os.path.join(val_path, class_name)
    os.makedirs(val_class_dir, exist_ok=True)

    for img in images[:split_size]:
        shutil.move(os.path.join(class_dir, img),
                    os.path.join(val_class_dir, img))

print("Validation Set Created ✅")
```

```python
import os
import shutil

BASE_PATH = "/content/SkinDisease"
PROCESSED_PATH = "/content/processed_dataset"

# Remove old processed folder if exists
shutil.rmtree(PROCESSED_PATH, ignore_errors=True)

# Copy train and test
shutil.copytree(os.path.join(BASE_PATH, "train"),
                os.path.join(PROCESSED_PATH, "train"))

shutil.copytree(os.path.join(BASE_PATH, "test"),
                os.path.join(PROCESSED_PATH, "test"))

print("Train and Test copied ✅")
```

```python
from sklearn.model_selection import train_test_split

train_dir = "/content/processed_dataset/train"
validation_dir = "/content/processed_dataset/validation"

os.makedirs(validation_dir, exist_ok=True)

for disease in os.listdir(train_dir):

    disease_path = os.path.join(train_dir, disease)
    images = os.listdir(disease_path)

    train_imgs, val_imgs = train_test_split(
        images,
        test_size=0.2,
        random_state=42
    )

    val_disease_path = os.path.join(validation_dir, disease)
    os.makedirs(val_disease_path, exist_ok=True)

    print(f"Creating validation data for {disease}")

    for img in val_imgs:
        src = os.path.join(disease_path, img)
        dst = os.path.join(val_disease_path, img)

        shutil.move(src, dst)

print("✅ Validation dataset created!")
```

```python
import cv2

def is_blurry(image_path, threshold=100):
    img = cv2.imread(image_path)
    if img is None:
        return True
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var() < threshold
```

```python
removed = 0

for split in ["train", "validation", "test"]:
    split_path = os.path.join(PROCESSED_PATH, split)

    for root, dirs, files in os.walk(split_path):
        for file in files:
            if file.lower().endswith((".jpg", ".png", ".jpeg")):
                path = os.path.join(root, file)
                if is_blurry(path, threshold=100):
                    os.remove(path)
                    removed += 1

print("Blurred images removed:", removed)
```

```python
import matplotlib.pyplot as plt
import random

def show_tiles(folder, n=9):
    imgs = []
    for root, dirs, files in os.walk(folder):
        for f in files:
            if f.lower().endswith((".jpg", ".png", ".jpeg")):
                imgs.append(os.path.join(root, f))

    sample = random.sample(imgs, min(n, len(imgs)))

    plt.figure(figsize=(8,8))
    for i, img_path in enumerate(sample):
        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.subplot(3,3,i+1)
        plt.imshow(img)
        plt.axis("off")
    plt.show()
```
